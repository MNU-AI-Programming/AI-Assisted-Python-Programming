## 준비: 자동 확인 도우미

아래 셀을 **먼저 실행**하세요. `check(설명, 검사함수)`는 검사 결과를 ✅/❌/⏳로 보여 줍니다.

````python
def check(desc, test):
    """자동 확인 도우미: test는 True/False를 돌려주는 함수(lambda)입니다."""
    try:
        ok = test()
    except NotImplementedError:
        print(f"⏳ 미작성 - {desc}")
    except Exception as e:
        print(f"❌ 오류 - {desc} ({type(e).__name__}: {e})")
    else:
        print(f"{'✅ 통과' if ok else '❌ 실패'} - {desc}")
````

```python
study_summary(15, 2)  # "총 30시간 (1,800분)"
study_summary(15, 3)  # "총 45시간 (2,700분)"
```

그런데 함수 안에 **버그가 3개** 있습니다. 실행해서 오류 메시지를 읽고 모두 고치세요. 그리고 아래 **버그 기록표**에 무엇을 왜 고쳤는지 적습니다.

````python
def study_summary(weeks, hours_per_week):
    """총 수업 시간을 '총 N시간 (M분)' 형식으로 돌려줍니다. 분은 천 단위 쉼표를 넣습니다."""
    total_hours = weeks * hours_per_weeks
    total_minutes = total_hours * 60
    message = "총 " + total_hours + "시간 (" + f"{total_minutes}" + "분)"
    print(message)
````

````python
check("study_summary(15, 2)", lambda: study_summary(15, 2) == "총 30시간 (1,800분)")
check("study_summary(15, 3)", lambda: study_summary(15, 3) == "총 45시간 (2,700분)")
check("study_summary(1, 1)", lambda: study_summary(1, 1) == "총 1시간 (60분)")
````

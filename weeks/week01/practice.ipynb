## 2. 오류 메시지 읽기

오류는 실패가 아니라 **Python이 주는 힌트**입니다. 오류 메시지는 다음 순서로 읽습니다.

1. **마지막 줄**을 먼저 봅니다. → `오류이름: 설명`
2. 오류 이름으로 종류를 파악합니다.
3. 위쪽의 `---->` 화살표나 `line N`으로 **위치**를 찾습니다.
4. 고친 뒤 다시 실행합니다.

| 오류 이름 | 흔한 원인 |
|---|---|
| `NameError` | 변수 이름 오타, 아직 실행하지 않은 셀의 변수 사용 |
| `TypeError` | 서로 맞지 않는 자료형끼리 연산 (`"3" + 4`) |
| `SyntaxError` | 괄호·따옴표·콜론 누락 등 문법 오류 |
| `ValueError` | 형식이 맞지 않는 값 변환 (`int("삼")`) |
| `IndentationError` | 들여쓰기가 맞지 않음 |

아래 `show_error()`는 코드를 실행해 보고 **오류가 나면 마지막 줄만** 보여 주는 도우미 함수입니다. 어떻게 만드는지는 6주차(예외 처리)에 배웁니다. 지금은 사용만 해 보세요.

````python
def show_error(code):
    """코드 문자열을 실행하고, 오류가 나면 오류 이름과 설명만 출력합니다."""
    try:
        exec(code, {})
        print("오류 없음 ✅")
    except Exception as e:
        print(f"{type(e).__name__}: {e}")

show_error('student = "김민준"\nprint(studnet)')
show_error('print("3" + 4)')
show_error('print("안녕"')
show_error('int("삼")')
````

### f-string 서식 지정

`{값:서식}` 형태로 자리 수, 소수점, 정렬을 지정할 수 있습니다.

| 서식 | 의미 | 예 |
|---|---|---|
| `:,` | 천 단위 쉼표 | `f"{1500000:,}"` → `1,500,000` |
| `:.2f` | 소수점 둘째 자리 | `f"{3.14159:.2f}"` → `3.14` |
| `:>8` | 8칸 오른쪽 정렬 | `f"{'A':>8}"` |
| `:<8` | 8칸 왼쪽 정렬 | `f"{'A':<8}"` |
| `:^8` | 8칸 가운데 정렬 | `f"{'A':^8}"` |

````python
weeks = 15
hours_per_week = 2
total_hours = weeks * hours_per_week
tuition = 1234500

print(f"{'총 수업 시간':<10}|{total_hours:>6}시간")
print(f"{'총 수업 분':<10}|{total_hours * 60:>6,}분")
print(f"{'등록금(예시)':<10}|{tuition:>10,}원")
print(f"{'주당 비율':<10}|{hours_per_week / total_hours:>8.2%}")
````

## 4. Level 3 적용하기: 수강 준비 진단 프로그램

이 수업을 듣기 위해 필요한 준비물을 점검하는 프로그램을 만들어 봅시다.

**요구사항**

1. 준비 항목 4개(Google 계정, GitHub 계정, 생성형 AI 사용 가능, Colab 실행 경험)를 `True/False`로 저장합니다.
2. 준비된 항목 수를 세고 백분율로 보여 줍니다.
3. 4개 모두 준비되면 "수강 준비 완료", 아니면 부족한 항목을 안내합니다.

> 💡 `True`는 숫자 1, `False`는 숫자 0처럼 더할 수 있습니다.

````python
has_google = True
has_github = False
can_use_ai = True
ran_colab = True

ready_count = has_google + has_github + can_use_ai + ran_colab
print(f"준비 항목: {ready_count}/4 ({ready_count / 4:.0%})")
````

````python
def diagnose(has_google, has_github, can_use_ai, ran_colab):
    """준비 상태를 받아 안내 문장을 돌려줍니다."""
    missing = ""
    if not has_google:
        missing += "Google 계정 "
    if not has_github:
        missing += "GitHub 계정 "
    if not can_use_ai:
        missing += "생성형 AI "
    if not ran_colab:
        missing += "Colab 실행 "

    if missing == "":
        return "✅ 수강 준비 완료"
    return f"⚠️ 준비 필요: {missing.strip()}"

print(diagnose(has_google, has_github, can_use_ai, ran_colab))
````

### 🧪 여러 경우로 검증하기

프로그램은 **한 번 잘 동작했다고 끝이 아닙니다.** 정상 사례, 경계 사례(모두 준비/모두 미준비)를 함께 실행해 봅니다.

````python
print("모두 준비   :", diagnose(True, True, True, True))
print("모두 미준비 :", diagnose(False, False, False, False))
print("GitHub만 없음:", diagnose(True, False, True, True))
````

`assert`는 **조건이 참이 아니면 멈추는** 검사 문장입니다. 기대 결과를 코드로 적어 두면, 나중에 코드를 고쳐도 자동으로 확인할 수 있습니다.

````python
assert diagnose(True, True, True, True) == "✅ 수강 준비 완료"
assert "GitHub 계정" in diagnose(True, False, True, True)
assert diagnose(False, False, False, False).startswith("⚠️")
print("모든 검사를 통과했습니다 ✅")
````

```text
- 정상 입력 1개, 경계 입력 1개, 잘못된 입력 1개의 테스트 예시도 보여 주세요.
```

### 🤖 AI가 준 코드 검증하기

어떤 학생이 "나이 계산 코드 줘"라고만 묻고 아래 코드를 받았습니다. 그대로 붙여 넣고 실행했더니 오류가 났습니다.

```python
birth = input("태어난 해: ")
age = 2026 - birth
print("나이는 " + age + "살")
```

`input()`은 문자열을 돌려준다는 점을 떠올리며 실행해 봅시다. (`input` 대신 문자열 `"2005"`를 넣었습니다.)

````python
show_error('birth = "2005"\nage = 2026 - birth\nprint("나이는 " + age + "살")')
````

**오류 분석**: `2026 - "2005"` → 숫자에서 문자열을 뺄 수 없어 `TypeError`가 납니다. 고치고 나면 `"나이는 " + 21`에서 또 `TypeError`가 납니다.

**수정 방법**: 입력을 `int()`로 바꾸고, 출력은 f-string을 씁니다. 그리고 함수로 만들어 여러 값으로 검증합니다.

````python
def calc_age(birth_year_text, base_year=2026):
    """태어난 해(문자열)를 받아 base_year 기준 나이(연 나이)를 돌려줍니다."""
    birth_year = int(birth_year_text.strip())
    return base_year - birth_year

print(calc_age("2005"))     # 정상 사례
print(calc_age(" 2026 "))   # 경계 사례: 공백 포함, 올해 출생
show_error('calc_age("이천오")')  # 잘못된 입력 → 아직은 오류 (6주차에 처리)
````

### ✅ AI 코드 검증 루틴 (매주 사용)

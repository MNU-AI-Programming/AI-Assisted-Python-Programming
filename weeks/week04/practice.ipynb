## 1. Level 1 따라하기: 리스트

````python
scores = [72, 95, 88, 59]
scores.append(100)        # 끝에 추가
scores.insert(0, 67)      # 0번 자리에 추가
print(scores)
print("길이", len(scores), "합", sum(scores), "최소", min(scores), "최대", max(scores))
print("앞 3개", scores[:3], "/ 뒤 2개", scores[-2:])
print("88 있나?", 88 in scores, "/ 88의 위치", scores.index(88))
````

````python
removed = scores.pop()    # 마지막 원소를 꺼내며 삭제
scores.remove(59)         # 값으로 삭제 (첫 번째 하나만)
print("꺼낸 값:", removed, "/ 남은 리스트:", scores)
````

### `sort()`와 `sorted()`의 차이

- `리스트.sort()`: **원본을 정렬**하고 `None`을 돌려줍니다.
- `sorted(리스트)`: 원본은 그대로 두고 **정렬된 새 리스트**를 돌려줍니다.

````python
data = [3, 1, 2]
new = sorted(data, reverse=True)
print("sorted 결과:", new, "/ 원본:", data)

result = data.sort()
print("sort 결과:", result, "/ 원본:", data)   # sort()는 None을 돌려준다!
````

### ❓ 예상해 보기: 같은 리스트를 가리키는 두 이름

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
```

````python
a = [1, 2, 3]
b = a            # 복사가 아니라 같은 리스트에 이름을 하나 더 붙임
b.append(4)
print("a:", a, "/ a is b:", a is b)

c = a.copy()     # 새 리스트로 복사 (list(a), a[:] 도 같음)
c.append(5)
print("a:", a, "/ c:", c)
````

### 리스트 컴프리헨션

`[식 for 변수 in 모음 if 조건]` — 반복 + 조건 + 모으기를 한 줄로 씁니다.

````python
scores = [72, 95, 88, 59, 100, 67, 81]

passed = [s for s in scores if s >= 70]
curved = [min(s + 5, 100) for s in scores]
labels = ["합격" if s >= 70 else "재시험" for s in scores]

print(passed)
print(curved)
print(labels)
````

## 2. 튜플: 바뀌지 않는 레코드

````python
student = ("김민준", "컴퓨터공학과", 2)   # (이름, 학과, 학년)
name, dept, year = student             # 언패킹
print(name, dept, year)

x, y = 10, 20
x, y = y, x                            # 두 값 바꾸기
print(x, y)

def min_max(values):
    return min(values), max(values)    # 여러 값 반환 = 튜플 반환

low, high = min_max([72, 95, 88])
print(low, high)
````

## 3. 집합: 중복 제거와 집합 연산

| 연산 | 기호 | 의미 |
|---|---|---|
| 합집합 | `a \| b` | 둘 중 하나라도 |
| 교집합 | `a & b` | 둘 다 |
| 차집합 | `a - b` | a에만 |
| 대칭차 | `a ^ b` | 한쪽에만 |

````python
submitted = ["김민준", "이서연", "김민준", "박지호", "이서연"]
unique = set(submitted)
print("제출 기록", len(submitted), "건 → 제출한 사람", len(unique), "명:", sorted(unique))
````

````python
python_class = {"김민준", "이서연", "박지호", "최하은"}
data_class = {"이서연", "최하은", "정도윤", "강지우"}

print("두 과목 모두:", sorted(python_class & data_class))
print("하나라도   :", sorted(python_class | data_class))
print("Python만   :", sorted(python_class - data_class))
print("한쪽만     :", sorted(python_class ^ data_class))
````

## 4. 딕셔너리: 이름으로 찾기

````python
student = {"name": "김민준", "dept": "컴퓨터공학과", "midterm": 78, "final": 85}

print(student["name"])
print(student.get("assignment"))          # 없는 키 → None
print(student.get("assignment", 0))       # 없는 키 → 기본값
student["assignment"] = 90                 # 추가
student["final"] = 88                      # 수정

for key, value in student.items():
    print(f"{key:>10}: {value}")
````

### 패턴 1: 개수 세기

`counts[key] = counts.get(key, 0) + 1` 한 줄이 핵심입니다.

````python
answers = ["예", "아니오", "예", "예", "모름", "아니오", "예"]

counts = {}
for a in answers:
    counts[a] = counts.get(a, 0) + 1
print(counts)

from collections import Counter     # 같은 일을 하는 표준 라이브러리
print(Counter(answers).most_common(2))
````

### 패턴 2: 그룹 묶기

키별로 **리스트를 값으로** 모읍니다. `setdefault(key, [])`는 키가 없으면 빈 리스트를 만들어 줍니다.

````python
pairs = [("컴퓨터공학과", 78), ("경영학과", 91), ("컴퓨터공학과", 85), ("간호학과", 88), ("경영학과", 70)]

groups = {}
for dept, score in pairs:
    groups.setdefault(dept, []).append(score)
print(groups)

for dept, s_list in groups.items():
    print(f"{dept}: {len(s_list)}명, 평균 {sum(s_list) / len(s_list):.1f}")
````

### `sorted`의 `key`: 무엇을 기준으로 정렬할까?

`key=lambda r: ...`는 "각 원소를 무엇으로 비교할지"를 알려 주는 작은 함수입니다. 튜플을 돌려주면 **여러 기준**으로 정렬할 수 있습니다.

````python
records = [("김민준", 85), ("이서연", 92), ("박지호", 85), ("최하은", 78)]

print(sorted(records, key=lambda r: r[1]))                    # 점수 오름차순
print(sorted(records, key=lambda r: r[1], reverse=True))      # 점수 내림차순
print(sorted(records, key=lambda r: (-r[1], r[0])))           # 점수 내림차순, 같으면 이름순
````

## 6. Level 3 적용하기: 학생 성적 레코드 분석

실제 데이터는 대부분 **"딕셔너리의 리스트"** 모양입니다. (한 학생 = 딕셔너리 하나, 전체 = 리스트)
6주차에는 같은 데이터를 CSV 파일에서 읽어 옵니다.

````python
students = [
    {"name": "김민준", "dept": "컴퓨터공학과", "year": 2, "midterm": 78, "final": 85, "attendance": 15},
    {"name": "이서연", "dept": "정보보안학과", "year": 3, "midterm": 92, "final": 88, "attendance": 16},
    {"name": "박지호", "dept": "컴퓨터공학과", "year": 1, "midterm": 65, "final": 70, "attendance": 11},
    {"name": "최하은", "dept": "경영학과", "year": 2, "midterm": 88, "final": 91, "attendance": 16},
    {"name": "정도윤", "dept": "정보보안학과", "year": 4, "midterm": 71, "final": 64, "attendance": 13},
    {"name": "강지우", "dept": "컴퓨터공학과", "year": 3, "midterm": 95, "final": 93, "attendance": 16},
    {"name": "조예준", "dept": "경영학과", "year": 1, "midterm": 58, "final": 66, "attendance": 10},
    {"name": "윤수아", "dept": "간호학과", "year": 2, "midterm": 84, "final": 90, "attendance": 15},
]
print(len(students), "명")
````

### 6-1. 파생 값 만들기: 총점 (중간 40% + 기말 60%)

원본을 바꾸지 않고, 필요한 값만 담은 **새 레코드**를 만듭니다.

````python
rows = []
for s in students:
    total = round(s["midterm"] * 0.4 + s["final"] * 0.6, 1)
    rows.append({"name": s["name"], "dept": s["dept"], "total": total, "attendance": s["attendance"]})

for r in rows[:3]:
    print(r)
````

### 6-2. 학과별 인원·평균·최고 학생

````python
by_dept = {}
for r in rows:
    by_dept.setdefault(r["dept"], []).append(r)

print(f"{'학과':<8}{'인원':>4}{'평균':>8}  최고 득점자")
for dept, members in sorted(by_dept.items()):
    avg = sum(m["total"] for m in members) / len(members)
    best = max(members, key=lambda m: m["total"])
    print(f"{dept:<8}{len(members):>4}{avg:>8.1f}  {best['name']}({best['total']})")
````

### 6-3. 상위 3명과 출석 미달자

- 상위 3명: 총점 내림차순, 같으면 이름순
- 출석 미달: 16회 중 12회 미만 (75% 미만)

````python
top3 = sorted(rows, key=lambda r: (-r["total"], r["name"]))[:3]
print("상위 3명:", [(r["name"], r["total"]) for r in top3])

low_attendance = {r["name"] for r in rows if r["attendance"] < 12}   # 집합 컴프리헨션
print("출석 미달:", sorted(low_attendance))
````

````python
# 검증: 요약 결과가 원본과 맞는가?
assert sum(len(m) for m in by_dept.values()) == len(students)        # 빠진 학생 없음
assert top3[0]["total"] == max(r["total"] for r in rows)              # 1등이 진짜 최고점
assert all(r["attendance"] < 12 for r in rows if r["name"] in low_attendance)
assert students[0]["midterm"] == 78                                   # 원본은 그대로
print("검증 통과 ✅")
````

## 7. Level 4 AI와 함께: 컬렉션 코드의 함정

### 함정 1: 반복하면서 리스트에서 삭제하기

"60점 미만을 리스트에서 지워 줘"라고 했더니 AI가 다음 코드를 주었습니다.

````python
scores = [55, 58, 90, 45, 70]

for s in scores:
    if s < 60:
        scores.remove(s)

print(scores)    # 기대: [90, 70]
````

**결과가 이상합니다.** 반복하는 도중에 원소를 지우면 뒤 원소가 한 칸씩 당겨져 **다음 원소를 건너뜁니다.** (55를 지우면 58이 0번 자리로 와서 검사되지 않음)

**해결**: 지우는 대신 **남길 것만 모아 새 리스트**를 만듭니다. 원본도 보존됩니다.

````python
scores = [55, 58, 90, 45, 70]
kept = [s for s in scores if s >= 60]
print("남긴 점수:", kept, "/ 원본:", scores)
````

### 함정 2: 없는 키에 더하기

````python
answers = ["예", "아니오", "예"]
counts = {}
try:
    for a in answers:
        counts[a] += 1          # AI가 자주 쓰는 실수
except KeyError as e:
    print("KeyError:", e)

counts = {}
for a in answers:
    counts[a] = counts.get(a, 0) + 1
print(counts)
````

### 함정 3: 함수가 몰래 원본을 바꾸기

````python
def add_bonus(score_list, bonus):
    for i in range(len(score_list)):
        score_list[i] += bonus          # 원본 리스트를 직접 수정
    return score_list

original = [70, 80, 90]
curved = add_bonus(original, 5)
print("curved:", curved, "/ original:", original)   # original도 바뀌었다!

def add_bonus_safe(score_list, bonus):
    return [s + bonus for s in score_list]           # 새 리스트를 만들어 반환

original = [70, 80, 90]
curved = add_bonus_safe(original, 5)
print("curved:", curved, "/ original:", original)
````

### 📝 AI 코드를 받을 때 확인할 것

| 확인 질문 | 테스트 방법 |
|---|---|
| 반복 중에 리스트를 지우거나 추가하지 않는가? | 연속된 삭제 대상 `[55, 58, ...]`을 넣어 본다 |
| 없는 키를 바로 꺼내지 않는가? | 처음 등장하는 키가 있는 입력으로 실행 |
| 원본을 바꾸지 않는가? | 실행 후 원본을 다시 출력하거나 `assert` |
| 동점은 어떻게 처리하는가? | 같은 점수 두 명을 넣어 본다 |

프롬프트에 **"원본 리스트는 바꾸지 말고 새 리스트를 반환"**, **"동점이면 이름 가나다순"** 처럼 적으면 이런 문제가 줄어듭니다.

## 8. 정리

| 오늘 배운 것 | 핵심 |
|---|---|
| 리스트 | `append` `pop` `remove`, 슬라이싱, `sorted` vs `sort()`, 복사는 `.copy()` |
| 컴프리헨션 | `[식 for x in 모음 if 조건]`, 집합·딕셔너리도 가능 |
| 튜플 | 바뀌지 않는 레코드, 언패킹, 여러 값 반환 |
| 집합 | 중복 제거, `&` `\|` `-` `^` |
| 딕셔너리 | `get(k, 기본값)`, `items()`, 개수 세기, `setdefault` 그룹 묶기 |
| 정렬 | `key=lambda r: (-r["total"], r["name"])` |
| AI 검증 | 반복 중 삭제, 없는 키, 원본 변경, 동점 처리 |

- [ ] 점수 내림차순·이름 오름차순 정렬의 `key`를 쓸 수 있다.

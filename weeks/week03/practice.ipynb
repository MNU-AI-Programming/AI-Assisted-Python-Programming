## 1. Level 1 따라하기: 조건문

| 비교 연산자 | 의미 | 논리 연산자 | 의미 |
|---|---|---|---|
| `==` `!=` | 같다 / 다르다 | `and` | 둘 다 참 |
| `<` `<=` | 작다 / 작거나 같다 | `or` | 하나라도 참 |
| `>` `>=` | 크다 / 크거나 같다 | `not` | 참·거짓 뒤집기 |

Python은 `80 <= score < 90`처럼 **비교를 이어 쓸 수** 있습니다.

````python
score = 85
print(score >= 80)
print(80 <= score < 90)
print(score >= 90 or score == 85)
print(not score > 90)
````

````python
temperature = 31

if temperature >= 33:
    print("폭염 경보: 실외 활동 자제")
elif temperature >= 28:
    print("더움: 물을 자주 마시세요")
elif temperature >= 10:
    print("활동하기 좋은 날씨")
else:
    print("추움: 따뜻하게 입으세요")
````

### ❓ 예상해 보기: 조건의 순서

아래 두 함수는 **조건의 순서만** 다릅니다. `score = 95`일 때 각각 무엇을 출력할까요?

````python
def grade_v1(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "F"


def grade_v2(score):
    if score >= 70:
        return "C"
    elif score >= 80:
        return "B"
    elif score >= 90:
        return "A"
    else:
        return "F"


print("v1:", grade_v1(95), "/ v2:", grade_v2(95))
````

`if-elif`는 **위에서부터 처음 참인 조건 하나만** 실행합니다. `95 >= 70`이 먼저 참이 되어 v2는 영원히 A를 줄 수 없습니다.
→ **범위가 좁은(엄격한) 조건을 먼저** 쓰거나, `80 <= score < 90`처럼 범위를 정확히 씁니다.

### 참·거짓으로 취급되는 값 (truthiness)

`0`, `0.0`, `""`(빈 문자열), `[]`(빈 리스트), `None`은 **거짓**처럼 취급됩니다.

````python
for value in [0, 7, "", "목포", [], [1, 2], None]:
    print(f"{value!r:10} → {'참' if value else '거짓'}")
````

````python
name = "   "
if not name.strip():
    print("이름이 비어 있습니다.")
````

### 중첩 조건 줄이기

조건 안에 조건이 겹겹이 쌓이면 읽기 어렵습니다. `and`로 합치거나, **안 되는 경우를 먼저 걸러 내면(early return)** 평평해집니다.

````python
def can_register_nested(year, credits, has_debt):
    if year >= 1:
        if credits <= 21:
            if not has_debt:
                return "신청 가능"
            else:
                return "미납금 있음"
        else:
            return "학점 초과"
    else:
        return "학년 오류"


def can_register_flat(year, credits, has_debt):
    if year < 1:
        return "학년 오류"
    if credits > 21:
        return "학점 초과"
    if has_debt:
        return "미납금 있음"
    return "신청 가능"


for args in [(2, 18, False), (2, 24, False), (3, 18, True), (0, 18, False)]:
    assert can_register_nested(*args) == can_register_flat(*args)
    print(args, "→", can_register_flat(*args))
````

## 2. Level 1 따라하기: 반복문

| 형태 | 언제 쓰나 |
|---|---|
| `for x in 모음:` | 모음(문자열·리스트 등)의 원소를 하나씩 |
| `for i in range(시작, 끝, 간격):` | 정해진 횟수만큼 (끝은 **미만**) |
| `for i, x in enumerate(모음, 1):` | 번호와 원소를 함께 |
| `while 조건:` | 조건이 참인 동안 (횟수를 모를 때) |

````python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -3)))
````

````python
subjects = ["Python", "자료구조", "데이터분석"]
for i, subject in enumerate(subjects, 1):
    print(f"{i}교시: {subject}")
````

### 반복 패턴 4가지

거의 모든 반복 문제는 다음 패턴의 조합입니다.

| 패턴 | 시작값 | 반복 안에서 |
|---|---|---|
| 누적(합계) | `total = 0` | `total += x` |
| 개수 세기 | `count = 0` | `if 조건: count += 1` |
| 최댓값 | `best = 첫 원소` (또는 `None`) | `if x > best: best = x` |
| 모으기 | `result = []` | `if 조건: result.append(x)` |

````python
scores = [72, 95, 88, 59, 100, 67, 81]

total = 0
passed = 0
best = scores[0]
low_scores = []

for s in scores:
    total += s
    if s >= 60:
        passed += 1
    if s > best:
        best = s
    if s < 70:
        low_scores.append(s)

print(f"평균 {total / len(scores):.1f}점, 합격 {passed}명, 최고 {best}점, 70점 미만 {low_scores}")
````

### `break`와 `continue`

- `continue`: 이번 반복만 건너뛰고 다음으로
- `break`: 반복 전체를 즉시 끝냄

````python
readings = [21.5, 22.0, -999, 23.1, 22.8, 85.0, 23.0]   # -999는 센서 오류, 85.0은 이상 고온

for t in readings:
    if t == -999:
        print("  센서 오류값 건너뜀")
        continue
    if t > 50:
        print(f"  ⚠️ 이상 고온 {t}도 감지 → 측정 중단")
        break
    print(f"정상 측정: {t}도")
````

### `while`: 몇 번 반복할지 모를 때

매달 30만 원씩 저축하면 몇 달 뒤에 200만 원 노트북을 살 수 있을까요?

> ⚠️ `while`은 조건이 **언젠가 거짓**이 되어야 끝납니다. 조건이 바뀌지 않으면 **무한 반복**이 됩니다. 안전장치(최대 횟수)를 두는 습관을 들이세요.

````python
goal = 2_000_000
monthly = 300_000
balance = 0
month = 0

while balance < goal:
    month += 1
    balance += monthly
    if month > 120:          # 안전장치: 10년이 넘으면 중단
        print("목표에 도달할 수 없습니다.")
        break

print(f"{month}개월 후 잔액 {balance:,}원")
````

## 4. Level 3 적용하기: 장학 기준 판정기

**장학 규정 (가상)**

| 장학 | 조건 | 우선순위 |
|---|---|---|
| 성적장학 | 평점 4.0 이상 **그리고** 이수학점 15 이상 | 1 |
| 성실장학 | 출석률 95% 이상 **그리고** 평점 3.0 이상 | 2 |
| 해당없음 | 위 조건 모두 아님 | – |

두 장학 조건을 모두 만족하면 **우선순위가 높은 성적장학 하나만** 받습니다.

### 4-1. 한 명 판정하기

````python
def scholarship(gpa, credits, attendance):
    """평점(4.5 만점), 이수학점, 출석률(%)로 장학 종류를 판정합니다."""
    if gpa >= 4.0 and credits >= 15:
        return "성적장학"
    if attendance >= 95 and gpa >= 3.0:
        return "성실장학"
    return "해당없음"


print(scholarship(4.2, 18, 90))
print(scholarship(3.5, 18, 97))
print(scholarship(4.3, 12, 99))
````

### 4-2. 경계값으로 검증하기

조건이 `>=`인지 `>`인지는 **정확히 경계에 있는 값**으로만 확인할 수 있습니다.

````python
cases = [
    # (gpa, credits, attendance, 기대 결과, 설명)
    (4.0, 15, 0, "성적장학", "성적 경계: 4.0, 15학점"),
    (3.99, 15, 0, "해당없음", "평점 경계 바로 아래"),
    (4.0, 14, 0, "해당없음", "학점 경계 바로 아래"),
    (3.0, 0, 95, "성실장학", "성실 경계: 3.0, 95%"),
    (2.99, 0, 100, "해당없음", "평점 경계 바로 아래"),
    (4.5, 21, 100, "성적장학", "둘 다 만족 → 우선순위"),
]

for gpa, credits, att, expected, desc in cases:
    result = scholarship(gpa, credits, att)
    mark = "✅" if result == expected else "❌"
    print(f"{mark} {desc:<22} → {result}")
````

### 4-3. 여러 학생 판정하고 요약하기

````python
students = [
    ("김민준", 4.21, 18, 92),
    ("이서연", 3.45, 17, 98),
    ("박지호", 4.05, 12, 100),
    ("최하은", 2.95, 19, 99),
    ("정도윤", 4.48, 21, 96),
    ("강지우", 3.10, 15, 95),
    ("조예준", 3.80, 18, 88),
]

merit = diligence = none = 0
best_name, best_gpa = None, -1

print(f"{'번호':<4}{'이름':<6}{'평점':>6}{'학점':>6}{'출석':>6}  판정")
for i, (name, gpa, credits, att) in enumerate(students, 1):
    result = scholarship(gpa, credits, att)
    if result == "성적장학":
        merit += 1
    elif result == "성실장학":
        diligence += 1
    else:
        none += 1
    if gpa > best_gpa:
        best_name, best_gpa = name, gpa
    print(f"{i:<5}{name:<6}{gpa:>6.2f}{credits:>6}{att:>6}%  {result}")

print(f"\n성적장학 {merit}명 · 성실장학 {diligence}명 · 해당없음 {none}명 / 최고 평점 {best_name}({best_gpa})")
assert merit + diligence + none == len(students)
````

## 5. Level 4 AI와 함께: 경계값 표로 검증하기

"점수를 넣으면 A~F 학점을 돌려주는 함수"를 요청했더니 AI가 아래 코드를 주었습니다. 몇 가지 점수로 실행하면 잘 되는 것처럼 보입니다.

````python
def grade_ai(score):
    if score > 90:
        return "A"
    elif score > 80:
        return "B"
    elif score > 70:
        return "C"
    elif score > 60:
        return "D"
    else:
        return "F"

print(grade_ai(95), grade_ai(85), grade_ai(75), grade_ai(30))
````

### 🔎 경계값 표 만들기

학점 규칙이 "90점 이상 A, 80점 이상 B, … 60점 미만 F, 0~100점 밖은 오류"라면, 다음 값들을 꼭 확인해야 합니다.

- 각 경계값 **그 자체**: 90, 80, 70, 60
- 경계 **바로 아래**: 89, 59
- 범위의 끝: 0, 100
- **잘못된 입력**: -1, 101

````python
expected = {100: "A", 90: "A", 89: "B", 80: "B", 70: "C", 60: "D", 59: "F", 0: "F", -1: "오류", 101: "오류"}

fails = 0
for score, want in expected.items():
    got = grade_ai(score)
    ok = got == want
    fails += not ok
    print(f"{'✅' if ok else '❌'} {score:>4}점 기대 {want:<3} 결과 {got}")
print(f"\n실패 {fails}건")
````

### 🛠️ 직접 고치기

- `>` → `>=` (경계값 포함)
- 입력 범위 검사를 **맨 앞에** 추가

````python
def grade_fixed(score):
    if score < 0 or score > 100:
        return "오류"
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 70:
        return "C"
    if score >= 60:
        return "D"
    return "F"


assert all(grade_fixed(s) == w for s, w in expected.items())
print("경계값 표 10개 모두 통과 ✅")
````

```text
점수(0~100 정수 또는 실수)를 받아 학점을 돌려주는 함수를 작성해 주세요.
- 90 이상 A, 80 이상 B, 70 이상 C, 60 이상 D, 60 미만 F
- 0 미만 또는 100 초과는 "오류"

```

AI는 **"이상/초과"** 를 자주 헷갈립니다. 경계값 표는 그런 실수를 잡는 가장 빠른 방법입니다.

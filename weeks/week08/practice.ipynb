````python
# 1. (2주) " 2026-09-07 " 형식의 날짜 문자열에서 월과 일을 정수로 꺼내 (월, 일) 튜플을 돌려주세요.
def month_day(text):
    # 직접 작성 ↓ (아래는 정답 예시입니다. 지우고 다시 써 보세요)
    _, m, d = text.strip().split("-")
    return int(m), int(d)

assert month_day(" 2026-09-07 ") == (9, 7)
assert month_day("2025-12-31") == (12, 31)
print("1번 ✅")
````

````python
# 2. (3주) 점수 리스트에서 60점 이상이 처음으로 3번 연속 나온 '마지막 점수의 위치(인덱스)'를 돌려주세요. 없으면 -1.
def first_streak3(scores):
    streak = 0
    for i, s in enumerate(scores):
        if s >= 60:
            streak += 1
            if streak == 3:
                return i
        else:
            streak = 0
    return -1

assert first_streak3([70, 50, 60, 61, 90, 40]) == 4
assert first_streak3([70, 70, 50, 70, 70]) == -1
assert first_streak3([]) == -1
print("2번 ✅")
````

````python
# 3. (4주) (학과, 점수) 리스트로 {학과: 최고점} 딕셔너리를 만드세요.
def best_by_dept(pairs):
    best = {}
    for dept, score in pairs:
        if dept not in best or score > best[dept]:
            best[dept] = score
    return best

assert best_by_dept([("컴공", 80), ("경영", 70), ("컴공", 95), ("경영", 65)]) == {"컴공": 95, "경영": 70}
assert best_by_dept([]) == {}
print("3번 ✅")
````

````python
# 4. (5주) 비율 문자열 "75%"를 0.75로 바꾸세요. 형식이 틀리거나 0~100% 밖이면 ValueError.
def parse_percent(text):
    text = text.strip()
    if not text.endswith("%"):
        raise ValueError(f"% 기호 없음: {text!r}")
    value = float(text[:-1])          # 숫자가 아니면 여기서 ValueError
    if not 0 <= value <= 100:
        raise ValueError(f"범위 밖: {value}")
    return value / 100

assert parse_percent(" 75% ") == 0.75
for bad in ["75", "abc%", "120%"]:
    try:
        parse_percent(bad)
        raise AssertionError(f"{bad}는 ValueError여야 함")
    except ValueError:
        pass
print("4번 ✅")
````

````python
# 5. (6주) CSV 텍스트를 파일로 저장한 뒤, csv.DictReader로 읽어 qty의 합을 구하세요. 숫자가 아닌 행은 건너뜁니다.
import csv
from pathlib import Path

Path("mini_orders.csv").write_text("menu,qty\n아메리카노,2\n라떼,abc\n베이글,1\n유자차, 3 \n", encoding="utf-8")

def total_qty(path):
    total = 0
    with open(path, encoding="utf-8", newline="") as f:
        for row in csv.DictReader(f):
            try:
                total += int(row["qty"])
            except ValueError:
                continue
    return total

assert total_qty("mini_orders.csv") == 6
print("5번 ✅")
````

````python
# 6. (7주) 주차장 클래스: 정원(capacity)을 넘으면 ValueError, 차량 번호 중복 입차도 ValueError.
class ParkingLot:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cars = set()

    def enter(self, plate):
        if plate in self.cars:
            raise ValueError(f"이미 입차: {plate}")
        if len(self.cars) >= self.capacity:
            raise ValueError("만차")
        self.cars.add(plate)

    def leave(self, plate):
        self.cars.remove(plate)          # 없으면 KeyError

    @property
    def free(self):
        return self.capacity - len(self.cars)

lot = ParkingLot(2)
lot.enter("12가3456"); lot.enter("34나7890")
assert lot.free == 0
try:
    lot.enter("56다1234"); raise AssertionError("만차여야 함")
except ValueError:
    pass
lot.leave("12가3456")
assert lot.free == 1
print("6번 ✅")
````

## 3. 모의 실기: 동아리 회비 관리

### 문제

동아리 총무가 회비 납부 기록 `club_fees.csv`를 관리합니다.

```text
date,member,amount,memo
2026-03-05,김민준,10000,3월 회비
...
```

1. **`load_fees(path)`**: 파일을 읽어 `(date, member, amount)` 기록 리스트와 건너뛴 줄 수를 돌려줍니다. 금액이 양의 정수가 아니거나 이름이 비어 있는 행은 건너뜁니다.
2. **`ClubLedger`** 클래스: 회원 명단과 납부 기록을 받아
   - `paid_total(member)`: 회원의 납부 합계
   - `unpaid(month, fee=10000)`: 해당 월(`"2026-03"`)에 `fee` 이상 내지 않은 회원 (가나다순)
   - `monthly_income()`: `{월: 합계}` (월 순서)
3. **`report(ledger, months)`**: 월별 수입과 미납자를 담은 보고서 문자열

### ① 요구사항 분석

| 구분 | 내용 |
|---|---|
| 입력 | CSV 파일 (date, member, amount, memo), 회원 명단 리스트 |
| 출력 | 기록 리스트, 회원별 합계, 월별 미납자, 월별 수입, 보고서 문자열 |
| 규칙 | 금액은 양의 정수, 날짜 앞 7글자가 월, 미납 = 그 달 합계 < 회비 |
| 예외 | 금액 숫자 아님/0 이하, 이름 빈칸 → 건너뛰고 개수 세기. 명단에 없는 납부자는 기록은 하되 미납 판정 대상 아님 |

### ② 설계

| 이름 | 종류 | 책임 |
|---|---|---|
| `load_fees(path)` | 함수 | 파일 읽기 + 행 검증 (6주차) |
| `ClubLedger.__init__(members, records)` | 메서드 | 명단·기록 보관 |
| `paid_total`, `unpaid`, `monthly_income` | 메서드 | 집계 (4주차 패턴) |
| `report(ledger, months)` | 함수 | 출력 문자열 만들기 (2주차 f-string) |

````python
# 모의 데이터 만들기
Path("club_fees.csv").write_text("""date,member,amount,memo
2026-03-05,김민준,10000,3월 회비
2026-03-06,이서연,10000,3월 회비
2026-03-09,박지호,5000,3월 일부
2026-03-20,박지호,5000,3월 나머지
2026-03-21,,10000,이름 누락
2026-04-02,김민준,10000,4월 회비
2026-04-03,이서연,만원,금액 오류
2026-04-10,최하은,10000,4월 회비
2026-04-11,외부인,20000,후원금
2026-04-15,박지호,-10000,환불 잘못 기록
""", encoding="utf-8")
MEMBERS = ["김민준", "이서연", "박지호", "최하은"]
````

### ③ 구현 (작은 것부터, 하나씩 실행)

````python
def load_fees(path):
    """(기록 리스트, 건너뛴 줄 수). 기록은 (date, member, amount) 튜플."""
    records, skipped = [], 0
    with open(path, encoding="utf-8", newline="") as f:
        for row in csv.DictReader(f):
            member = row["member"].strip()
            try:
                amount = int(row["amount"].strip())
            except ValueError:
                skipped += 1
                continue
            if not member or amount <= 0:
                skipped += 1
                continue
            records.append((row["date"].strip(), member, amount))
    return records, skipped


records, skipped = load_fees("club_fees.csv")
print(len(records), "건 읽음,", skipped, "건 건너뜀")
records[:3]
````

````python
class ClubLedger:
    def __init__(self, members, records):
        self.members = list(members)
        self.records = list(records)          # 원본 리스트를 복사해 보관

    def paid_total(self, member):
        return sum(amount for _d, m, amount in self.records if m == member)

    def paid_in_month(self, member, month):
        return sum(amount for d, m, amount in self.records if m == member and d[:7] == month)

    def unpaid(self, month, fee=10000):
        return sorted(m for m in self.members if self.paid_in_month(m, month) < fee)

    def monthly_income(self):
        income = {}
        for d, _m, amount in self.records:
            income[d[:7]] = income.get(d[:7], 0) + amount
        return dict(sorted(income.items()))


ledger = ClubLedger(MEMBERS, records)
print(ledger.paid_total("박지호"), ledger.unpaid("2026-03"), ledger.unpaid("2026-04"), ledger.monthly_income())
````

````python
def report(ledger, months):
    lines = ["[동아리 회비 보고서]"]
    income = ledger.monthly_income()
    for month in months:
        unpaid = ledger.unpaid(month)
        names = ", ".join(unpaid) if unpaid else "없음"
        lines.append(f"{month} 수입 {income.get(month, 0):,}원 / 미납 {len(unpaid)}명: {names}")
    return "\n".join(lines)


print(report(ledger, ["2026-03", "2026-04"]))
````

### ④ 테스트: 정상·경계·오류

요구사항 표의 **규칙과 예외 한 줄마다 테스트가 하나 이상** 있어야 합니다.

````python
def test_load_fees():
    recs, skipped = load_fees("club_fees.csv")
    assert skipped == 3                                    # 이름 누락, '만원', 음수
    assert ("2026-03-09", "박지호", 5000) in recs


def test_ledger():
    lg = ClubLedger(MEMBERS, load_fees("club_fees.csv")[0])
    assert lg.paid_total("박지호") == 10000                # 나누어 낸 금액 합산
    assert lg.unpaid("2026-03") == ["최하은"]               # 박지호는 합계 10000 → 납부
    assert lg.unpaid("2026-04") == ["박지호", "이서연"]      # 가나다순
    assert lg.unpaid("2026-05") == sorted(MEMBERS)          # 기록 없는 달 = 전원 미납
    assert lg.monthly_income() == {"2026-03": 30000, "2026-04": 40000}   # 외부인 후원금 포함


def test_ledger_does_not_mutate():
    recs = load_fees("club_fees.csv")[0]
    copy_before = list(recs)
    lg = ClubLedger(MEMBERS, recs)
    lg.records.append(("2026-05-01", "김민준", 1))
    assert recs == copy_before


def test_report_format():
    text = report(ClubLedger(MEMBERS, load_fees("club_fees.csv")[0]), ["2026-03"])
    assert text.splitlines() == ["[동아리 회비 보고서]", "2026-03 수입 30,000원 / 미납 1명: 최하은"]


for t in [test_load_fees, test_ledger, test_ledger_does_not_mutate, test_report_format]:
    t()
    print("✅", t.__name__)
````

### ⑤ 기술 메모 (예시)

> **설계**: 파일 읽기·검증(`load_fees`), 집계(`ClubLedger`), 출력(`report`)을 분리했다. 월은 날짜 앞 7글자로 판단해 날짜 형식이 같다는 전제를 둔다.
>
> **테스트**: 나누어 낸 납부(박지호 3월), 명단 밖 납부자(외부인), 기록 없는 달, 잘못된 금액 3종을 확인했다. 원본 기록 리스트가 바뀌지 않는지도 테스트했다.
>
> **AI 사용**: `monthly_income`의 월 정렬 방법을 AI에게 물어 `dict(sorted(...))`를 채택했다. AI가 처음 제안한 `except: continue`는 오타를 숨길 수 있어 `except ValueError`로 바꿨다.
>
> **한계**: 날짜 형식이 `YYYY-MM-DD`가 아니면 월 판단이 틀린다. 환불(음수)은 지금은 오류로 보고 건너뛰는데, 실제로는 별도 처리 규칙이 필요하다.

메모는 **무엇을 했는지**보다 **왜 그렇게 했는지, 무엇을 확인했는지, 무엇이 남았는지**를 쓰는 것이 핵심입니다.

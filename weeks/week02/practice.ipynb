## 1. Level 1 따라하기: 산술 연산

| 연산자 | 의미 | 예 | 결과 |
|---|---|---|---|
| `+` `-` `*` | 더하기·빼기·곱하기 | `7 * 3` | `21` |
| `/` | 나누기 (항상 실수) | `7 / 2` | `3.5` |
| `//` | 몫 (정수 나눗셈) | `7 // 2` | `3` |
| `%` | 나머지 | `7 % 2` | `1` |
| `**` | 거듭제곱 | `2 ** 10` | `1024` |

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(type(6 / 3))   # 나누어떨어져도 / 는 float
````

### ❓ 예상해 보기: 연산 우선순위

`**` → `*` `/` `//` `%` → `+` `-` 순서로 계산합니다. 헷갈리면 **괄호**를 씁니다.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(-2 ** 2)
print(10 - 4 // 3 * 2)
```

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(-2 ** 2)        # -(2**2) 로 계산됩니다
print(10 - 4 // 3 * 2)
````

### `//`와 `%`의 단짝 활용: 단위 바꾸기

몫과 나머지를 함께 쓰면 초 → 분·초, 원 → 만 원·천 원처럼 **큰 단위와 작은 단위**로 나눌 수 있습니다. `divmod(a, b)`는 몫과 나머지를 한 번에 돌려줍니다.

````python
total_seconds = 7384
hours = total_seconds // 3600
minutes = total_seconds % 3600 // 60
seconds = total_seconds % 60
print(f"{total_seconds}초 = {hours}시간 {minutes}분 {seconds}초")

q, r = divmod(total_seconds, 60)
print("divmod(7384, 60) =", (q, r))
````

````python
# 짝수/홀수, 배수 판별에도 % 를 씁니다.
n = 2026
print(f"{n}은 짝수인가? {n % 2 == 0}")
print(f"{n}은 4의 배수인가? {n % 4 == 0}")
print(f"{n}의 일의 자리: {n % 10}, 십의 자리: {n // 10 % 10}")
````

### ⚠️ 실수 계산의 함정

컴퓨터는 실수를 2진수로 저장하기 때문에 `0.1` 같은 값을 **정확히 표현하지 못합니다.**

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)

import math
print(math.isclose(0.1 + 0.2, 0.3))   # 실수 비교는 isclose로
````

### ❓ 예상해 보기: 반올림

`round(2.5)`는 3일까요? 실행 전에 예상해 보세요.

````python
print(round(2.5), round(3.5), round(4.5))
print(round(2.675, 2))
````

Python의 `round()`는 **은행가 반올림**(가운데 값이면 짝수 쪽으로)을 씁니다. 게다가 `2.675`는 실제로 `2.67499999...`로 저장되어 `2.67`이 됩니다.

돈 계산처럼 규칙이 정해진 경우에는 **올림·내림을 명시적으로** 쓰는 것이 안전합니다.

| 함수 | 의미 | 2.3을 넣으면 | -2.3을 넣으면 |
|---|---|---|---|
| `math.ceil` | 올림 | 3 | -2 |
| `math.floor` | 내림 | 2 | -3 |
| `int()` | 0 쪽으로 버림 | 2 | -2 |

````python
import math

price = 33333.4
print("올림 :", math.ceil(price))
print("내림 :", math.floor(price))
print("버림 :", int(price))

# 100원 단위로 올림하기: 단위로 나눈 뒤 올림하고 다시 곱한다
unit = 100
print("100원 단위 올림:", math.ceil(price / unit) * unit)
print("100원 단위 내림:", math.floor(price / unit) * unit)
````

 문자열:   2  0  2  6  1  2  0  3  4
 인덱스:   0  1  2  3  4  5  6  7  8
 음수  :  -9 -8 -7 -6 -5 -4 -3 -2 -1

### 문자열 메서드

문자열은 **바꿀 수 없는(immutable)** 값입니다. 메서드는 원본을 바꾸지 않고 **새 문자열을 돌려줍니다.**

| 메서드 | 기능 | 예 → 결과 |
|---|---|---|
| `strip()` | 앞뒤 공백 제거 | `"  hi "` → `"hi"` |
| `upper()` / `lower()` | 대/소문자 | `"Py"` → `"PY"` |
| `replace(a, b)` | a를 b로 바꿈 | `"1,000"` → `"1000"` |
| `split(sep)` | 나누어 리스트로 | `"a,b"` → `["a", "b"]` |
| `sep.join(list)` | 리스트를 이어 붙임 | `"-".join(["a","b"])` → `"a-b"` |
| `startswith` / `endswith` | 시작·끝 확인 | `"목포대".startswith("목포")` → `True` |
| `count(x)` / `find(x)` | 개수 / 위치 | `"banana".count("a")` → `3` |
| `isdigit()` | 숫자로만 이루어졌는가 | `"2026"` → `True` |

````python
raw = "   Mokpo National University   "
clean = raw.strip()
print(f"[{raw}]")
print(f"[{clean}]")
print(clean.upper())
print(clean.replace("National", "Natl."))
print(clean.split())        # 공백 기준으로 나누기
print(raw)                  # 원본은 그대로!
````

````python
price_text = "₩1,250,000"
number = int(price_text.replace("₩", "").replace(",", ""))
print(number, type(number))
print(f"{number * 1.1:,.0f}원 (부가세 포함)")
````

````python
cities = "  목포 , 여수,순천 ,  광주 "
parts = cities.split(",")
print("split 결과:", parts)

clean_parts = [p.strip() for p in parts]   # 각 조각의 공백 제거 (4주차에 자세히)
print("정리 결과:", clean_parts)
print(" → ".join(clean_parts))
````

````python
for text in ["2026", " 2026 ", "2026년", "-5", "3.5"]:
    print(f"{text!r:10} isdigit={text.isdigit()!s:5}  strip 후={text.strip().isdigit()}")
````

## 4. Level 3 적용하기: 여행 경비 계산기

좋은 프로그램은 **입력 → 처리 → 출력**이 분리되어 있습니다. 그래야 입력을 바꾸어 다시 검증하기 쉽습니다.

**요구사항**

1. 경비는 `"항목:금액"`을 쉼표로 이은 한 줄 문자열로 입력받습니다. 금액에는 쉼표가 있을 수 있습니다.
2. 항목별 금액과 총액을 영수증 형식으로 출력합니다.
3. 1인당 금액은 **100원 단위로 올림**합니다.

### 4-1. 입력

````python
USE_INPUT = False

if USE_INPUT:
    expense_text = input("경비 (항목:금액, 쉼표로 구분): ")
    people_text = input("인원: ")
else:
    expense_text = "숙박:120,000 / 교통:64,800 / 식비:183,500 / 입장료:24,000"
    people_text = "4"
````

### 4-2. 처리

쉼표가 금액 안에도 있기 때문에 항목 구분자로 `/`를 썼습니다. 이렇게 **구분자를 정하는 것**도 설계입니다.

````python
import math

items = []          # (항목, 금액) 쌍을 모을 리스트
total = 0
for chunk in expense_text.split("/"):
    name, amount_text = chunk.split(":")
    amount = int(amount_text.strip().replace(",", ""))
    items.append((name.strip(), amount))
    total += amount

people = int(people_text)
per_person = math.ceil(total / people / 100) * 100

print(items)
print("총액:", total, "/ 1인당:", per_person)
````

### 4-3. 출력: 영수증 만들기

````python
line = "-" * 28
print(f"{'여행 경비 영수증':^24}")
print(line)
for name, amount in items:
    print(f"{name:<8}{amount:>14,}원")
print(line)
print(f"{'합계':<8}{total:>14,}원")
print(f"{'인원':<8}{people:>14}명")
print(f"{'1인당(올림)':<8}{per_person:>11,}원")
print(line)
````

> 💡 한글은 화면에서 영문보다 넓게 보여 정렬이 살짝 어긋날 수 있습니다. 내용이 맞는지가 우선입니다.

### 4-4. 검증

1인당 금액을 올림했으니 **모두 보낸 금액이 총액 이상**이어야 하고, 남는 돈은 **인원 × 100원 미만**이어야 합니다.

````python
collected = per_person * people
surplus = collected - total
print(f"모은 금액 {collected:,}원 - 총액 {total:,}원 = 남는 돈 {surplus:,}원")

assert collected >= total
assert 0 <= surplus < people * 100
print("검증 통과 ✅")
````

## 5. Level 4 AI와 함께: 정산 코드 검증

한 학생이 AI에게 **"총액을 n명이 나눠 내는 정산 코드"** 를 요청해 아래 코드를 받았습니다.

```python
def split_bill(total, people):
    each = round(total / people, -2)   # 100원 단위 반올림
    return each
```

겉보기엔 그럴듯합니다. 정말 맞는지 **여러 값으로** 검증해 봅시다.

````python
def split_bill_ai(total, people):
    each = round(total / people, -2)   # 100원 단위 반올림
    return each

for total, people in [(100000, 4), (100000, 3), (392300, 4), (50000, 6)]:
    each = split_bill_ai(total, people)
    diff = each * people - total
    print(f"총액 {total:>7,} / {people}명 → 1인 {each:>9,} → 모은 돈과 차이 {diff:>+6,.0f}")
````

### 🔎 발견한 문제

1. 반환값이 `33300.0`처럼 **실수**입니다. 돈은 정수여야 합니다.
2. 반올림이라 **모은 돈이 총액보다 적을 수 있습니다** (`100000 / 3` → 33,300 × 3 = 99,900원 → 100원 부족).
3. 프롬프트에 **"부족하면 안 된다"는 규칙**을 주지 않았기 때문입니다.

### 🛠️ 개선: 규칙을 먼저 정하기

- 규칙 A: 모두 100원 단위로 **올림**해서 낸다 (남는 돈은 총무가 보관)
- 규칙 B: n−1명은 100원 단위로 **내림**한 금액을 내고, **마지막 사람이 나머지**를 낸다 → 합이 총액과 정확히 같다

````python
def split_bill_ceil(total, people, unit=100):
    """규칙 A: 1인당 금액을 unit 단위로 올림."""
    return math.ceil(total / people / unit) * unit


def split_bill_exact(total, people, unit=100):
    """규칙 B: (일반 1인 금액, 마지막 사람 금액)을 돌려준다. 합계 = total"""
    each = total // people // unit * unit
    last = total - each * (people - 1)
    return each, last


for total, people in [(100000, 4), (100000, 3), (392300, 4), (50000, 6)]:
    a = split_bill_ceil(total, people)
    each, last = split_bill_exact(total, people)
    print(f"{total:>7,}원/{people}명 | A: {a:,}원씩 | B: {each:,}원 × {people - 1} + 마지막 {last:,}원")
````

````python
# 자동 검증: 여러 경우에 대해 규칙이 지켜지는지 확인
for total in [1000, 99999, 100000, 392300, 1234567]:
    for people in [1, 2, 3, 4, 7]:
        a = split_bill_ceil(total, people)
        each, last = split_bill_exact(total, people)
        assert a * people >= total                      # A: 부족하지 않다
        assert each * (people - 1) + last == total      # B: 정확히 같다
        assert isinstance(each, int) and isinstance(last, int)
print("25가지 경우 모두 통과 ✅")
````

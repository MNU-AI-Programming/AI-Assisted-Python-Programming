## 준비: 자동 확인 도우미

`raises(예외종류, 함수, 인자...)`는 함수가 **그 예외를 발생시키면 True**를 돌려줍니다.

````python
import math

def check(desc, test):
    """자동 확인 도우미: test는 True/False를 돌려주는 함수(lambda)입니다."""
    try:
        ok = test()
    except NotImplementedError:
        print(f"⏳ 미작성 - {desc}")
    except Exception as e:
        print(f"❌ 오류 - {desc} ({type(e).__name__}: {e})")
    else:
        print(f"{'✅ 통과' if ok else '❌ 실패'} - {desc}")


def raises(exc_type, fn, *args):
    """fn(*args)가 exc_type 예외를 내면 True, 아니면 False."""
    try:
        fn(*args)
    except exc_type:
        return True
    except NotImplementedError:
        raise
    return False
````

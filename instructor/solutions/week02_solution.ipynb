## 문제 1. 초를 `HH:MM:SS`로 바꾸기 (2점)

0 이상의 정수 초 `sec`를 받아 `"HH:MM:SS"` 형식 문자열을 돌려주는 `to_hms(sec)`를 완성하세요.

- 시·분·초는 항상 **두 자리**로 씁니다. (`5` → `"05"`)
- 시간이 100 이상이면 그대로 씁니다. (`360000` → `"100:00:00"`)
- `//`, `%` 또는 `divmod`를 사용하세요.

````python
def to_hms(sec):
    """초를 'HH:MM:SS' 문자열로 바꿉니다."""
    h, rest = divmod(sec, 3600)
    m, s = divmod(rest, 60)
    return f"{h:02d}:{m:02d}:{s:02d}"
````

````python
check("to_hms(0) == '00:00:00'", lambda: to_hms(0) == "00:00:00")
check("to_hms(59) == '00:00:59'", lambda: to_hms(59) == "00:00:59")
check("to_hms(3600) == '01:00:00'", lambda: to_hms(3600) == "01:00:00")
check("to_hms(7384) == '02:03:04'", lambda: to_hms(7384) == "02:03:04")
check("to_hms(86399) == '23:59:59'", lambda: to_hms(86399) == "23:59:59")
check("to_hms(360000) == '100:00:00'", lambda: to_hms(360000) == "100:00:00")
````

## 문제 2. 학번 분석하기 (2점)

학번은 9자리 문자열입니다: **입학년도 4자리 + 학과코드 2자리 + 번호 3자리** (예: `"202612034"`)

`parse_student_id(sid)`는 다음 문자열을 돌려줍니다.

```python
parse_student_id("202612034")    # "2026학번 / 학과 12 / 34번"
parse_student_id(" 202503007 ")  # "2025학번 / 학과 03 / 7번"
```

- 앞뒤 공백은 제거합니다.
- 번호는 **앞의 0을 뺀 정수**로 씁니다. (`"007"` → `7`) 학과코드는 문자열 그대로 씁니다.

````python
def parse_student_id(sid):
    """학번 문자열을 '연도학번 / 학과 코드 / 번호' 형식으로 돌려줍니다."""
    sid = sid.strip()
    year = sid[:4]
    dept = sid[4:6]
    number = int(sid[6:])
    return f"{year}학번 / 학과 {dept} / {number}번"
````

````python
check("기본 학번", lambda: parse_student_id("202612034") == "2026학번 / 학과 12 / 34번")
check("앞뒤 공백 + 앞자리 0", lambda: parse_student_id(" 202503007 ") == "2025학번 / 학과 03 / 7번")
check("번호 100번 이상", lambda: parse_student_id("202499150") == "2024학번 / 학과 99 / 150번")
````

## 문제 3. 영수증 문자열 만들기 (3점)

카페 주문 내역 문자열을 받아 영수증 문자열을 돌려주는 `make_receipt(order_text)`를 완성하세요.

- 입력: `"메뉴:가격"`을 `/`로 이은 문자열. 공백과 천 단위 쉼표가 섞여 있을 수 있습니다.
- 출력: 메뉴마다 한 줄 `"메뉴: 가격원"`, 마지막 줄 `"합계: 총액원"`. 금액은 천 단위 쉼표를 넣고 줄은 `"\n"`으로 연결합니다.

```python
make_receipt("아메리카노:2,500 / 베이글: 3000")
# "아메리카노: 2,500원\n베이글: 3,000원\n합계: 5,500원"
```

````python
def make_receipt(order_text):
    """주문 문자열을 영수증 문자열로 바꿉니다."""
    lines = []
    total = 0
    for chunk in order_text.split("/"):
        name, price_text = chunk.split(":")
        price = int(price_text.strip().replace(",", ""))
        lines.append(f"{name.strip()}: {price:,}원")
        total += price
    lines.append(f"합계: {total:,}원")
    return "\n".join(lines)
````

````python
check("두 메뉴", lambda: make_receipt("아메리카노:2,500 / 베이글: 3000")
      == "아메리카노: 2,500원\n베이글: 3,000원\n합계: 5,500원")
check("한 메뉴", lambda: make_receipt("치즈케이크:4500") == "치즈케이크: 4,500원\n합계: 4,500원")
check("공백이 많은 입력", lambda: make_receipt("  카페라떼 : 3,200 /유자차:3,000  / 크루아상 :2,800 ")
      == "카페라떼: 3,200원\n유자차: 3,000원\n크루아상: 2,800원\n합계: 9,000원")
check("큰 금액", lambda: make_receipt("단체주문:1,250,000").endswith("합계: 1,250,000원"))
````

## 문제 4. 정확한 정산 함수 (AI 협업, 3점)

`split_exact(total, people, unit=100)`은 총액을 `people`명이 나눠 낼 때 **(일반 1인 금액, 마지막 사람 금액)** 을 돌려줍니다.

**규칙**

1. 일반 1인 금액은 `total / people`을 `unit` 단위로 **내림**한 정수입니다.
2. 마지막 사람은 나머지를 모두 냅니다. 따라서 `일반 × (people − 1) + 마지막 == total`이 **항상** 성립해야 합니다.
3. 결과는 모두 `int`입니다. `people`이 1이면 `(0, total)`이 아니라 `(일반, total)`이어도 되지만, 합계 규칙은 지켜야 합니다.

```python
split_exact(100000, 3)        # (33300, 33400)
split_exact(392300, 4)        # (98000, 98300)
split_exact(100000, 3, 1000)  # (33000, 34000)
```

**진행 순서**: 위 규칙을 담은 프롬프트로 AI에게 코드를 받고 → 자동 확인 실행 → 실패하면 원인을 찾아 직접 수정 → 기록.

````python
def split_exact(total, people, unit=100):
    """(일반 1인 금액, 마지막 사람 금액)을 돌려줍니다. 합계는 total과 정확히 같습니다."""
    each = total // people // unit * unit
    last = total - each * (people - 1)
    return each, last
````

````python
check("100000원 / 3명", lambda: split_exact(100000, 3) == (33300, 33400))
check("392300원 / 4명", lambda: split_exact(392300, 4) == (98000, 98300))
check("1000원 단위", lambda: split_exact(100000, 3, 1000) == (33000, 34000))
check("결과가 정수", lambda: all(isinstance(v, int) for v in split_exact(99999, 7)))


def _sum_rule_holds():
    for total in [100, 9999, 100000, 392300, 1234567]:
        for people in [1, 2, 3, 4, 7, 10]:
            each, last = split_exact(total, people)
            if each * (people - 1) + last != total or each % 100 != 0:
                return False
    return True

check("30가지 경우 합계 규칙", _sum_rule_holds)
````

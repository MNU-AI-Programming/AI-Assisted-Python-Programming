## 준비: 자동 확인 도우미

아래 셀을 **먼저 실행**하세요. `check(설명, 검사함수)`는 검사 결과를 ✅/❌/⏳로 보여 줍니다.

````python
def check(desc, test):
    """자동 확인 도우미: test는 True/False를 돌려주는 함수(lambda)입니다."""
    try:
        ok = test()
    except NotImplementedError:
        print(f"⏳ 미작성 - {desc}")
    except Exception as e:
        print(f"❌ 오류 - {desc} ({type(e).__name__}: {e})")
    else:
        print(f"{'✅ 통과' if ok else '❌ 실패'} - {desc}")
````

## 문제 1. 자기소개 변수 만들기 (2점)

다음 조건에 맞게 변수 4개를 만드세요.

| 변수 | 자료형 | 조건 |
|---|---|---|
| `name` | `str` | 비어 있지 않은 이름 |
| `major` | `str` | "과" 또는 "부"로 끝나는 소속 학과·학부 |

````python
name = "김목포"
major = "컴퓨터공학과"
year = 2
goal = "AI가 만든 코드를 직접 검증하는 습관 만들기"
````

## 문제 2. 온도 변환 함수 (2점)

섭씨(°C)와 화씨(°F)를 서로 바꾸는 함수 두 개를 완성하세요.

- 화씨 = 섭씨 × 9 / 5 + 32
- 섭씨 = (화씨 − 32) × 5 / 9
- 결과는 `round(값, 1)`로 **소수 첫째 자리**까지 반올림해 돌려줍니다.

```python
c_to_f(100)  # 212.0
f_to_c(212)  # 100.0
```

````python
def c_to_f(c):
    """섭씨 c를 화씨로 바꿔 소수 첫째 자리까지 돌려줍니다."""
    return round(c * 9 / 5 + 32, 1)


def f_to_c(f):
    """화씨 f를 섭씨로 바꿔 소수 첫째 자리까지 돌려줍니다."""
    return round((f - 32) * 5 / 9, 1)
````

````python
check("c_to_f(100) == 212.0", lambda: c_to_f(100) == 212.0)
check("c_to_f(0) == 32.0", lambda: c_to_f(0) == 32.0)
check("c_to_f(-40) == -40.0 (두 단위가 같아지는 온도)", lambda: c_to_f(-40) == -40.0)
check("f_to_c(98.6) == 37.0", lambda: f_to_c(98.6) == 37.0)
check("f_to_c(c_to_f(25)) == 25.0 (왕복 변환)", lambda: f_to_c(c_to_f(25)) == 25.0)
````

## 문제 3. 버그 고치기: 수강 시간 요약 (3점)

아래 함수는 주 수와 주당 수업 시간을 받아 다음 형식의 문자열을 **돌려주어야(return)** 합니다.

```python
study_summary(15, 2)  # "총 30시간 (1,800분)"
study_summary(15, 3)  # "총 45시간 (2,700분)"
```

그런데 함수 안에 **버그가 3개** 있습니다. 실행해서 오류 메시지를 읽고 모두 고치세요. 그리고 아래 **버그 기록표**에 무엇을 왜 고쳤는지 적습니다.

````python
def study_summary(weeks, hours_per_week):
    """총 수업 시간을 '총 N시간 (M분)' 형식으로 돌려줍니다. 분은 천 단위 쉼표를 넣습니다."""
    total_hours = weeks * hours_per_week
    total_minutes = total_hours * 60
    message = f"총 {total_hours}시간 ({total_minutes:,}분)"
    return message
````

````python
check("study_summary(15, 2)", lambda: study_summary(15, 2) == "총 30시간 (1,800분)")
check("study_summary(15, 3)", lambda: study_summary(15, 3) == "총 45시간 (2,700분)")
check("study_summary(1, 1)", lambda: study_summary(1, 1) == "총 1시간 (60분)")
````

### 버그 기록표 (예시 답안)

| 번호 | 오류 이름 또는 증상 | 원인 | 고친 방법 |
|---|---|---|---|
| 1 | `NameError: name 'hours_per_weeks' is not defined` | 매개변수 이름 오타 | `hours_per_week`로 수정 |
| 2 | `TypeError: can only concatenate str (not "int") to str` | 문자열과 정수를 `+`로 연결 | f-string으로 바꿈 |
| 3 | 결과가 `None`, 쉼표 없음 | `print`만 하고 `return`하지 않음, 서식 누락 | `return message`, `{total_minutes:,}` |

## 문제 4. AI와 함께 만드는 자기소개 카드 (3점)

생성형 AI를 활용해 `make_card(name, major, goal)` 함수를 완성하세요. 함수는 **출력하지 않고** 아래 5줄짜리 문자열을 **돌려줍니다**(`"\n"`으로 줄을 구분).

```text
==============================
이름: 김목포
전공: 컴퓨터공학과
목표: 매주 과제 제출하기
==============================
```

- 첫 줄과 마지막 줄은 `=` 30개입니다.
- 각 줄 끝에 공백이 없어야 하며, 이름·전공·목표 앞뒤 공백은 `strip()`으로 제거합니다.

**진행 순서**

1. 아래 칸에 4요소(역할·맥락 / 요구사항 / 제약 / 출력 형식)를 갖춘 프롬프트를 먼저 작성합니다.
2. AI의 답을 받아 붙여 넣고, **자동 확인으로 검증**한 뒤 필요한 부분을 직접 고칩니다.
3. AI 활용 기록을 작성합니다.

**프롬프트 예시**

```text
Python을 배우는 대학생이고 Google Colab을 씁니다.
make_card(name, major, goal) 함수를 만들어 주세요.
- 출력(print)하지 말고 문자열을 return 합니다.
- 5줄: '=' 30개 / '이름: {name}' / '전공: {major}' / '목표: {goal}' / '=' 30개
- 줄 구분은 '\n', 각 값의 앞뒤 공백은 strip()으로 제거합니다.
- f-string만 사용하고, 정상 입력과 공백이 섞인 입력으로 테스트 예시를 보여 주세요.
```

````python
def make_card(name, major, goal):
    """자기소개 카드 문자열(5줄)을 돌려줍니다."""
    border = "=" * 30
    lines = [
        border,
        f"이름: {name.strip()}",
        f"전공: {major.strip()}",
        f"목표: {goal.strip()}",
        border,
    ]
    return "\n".join(lines)
````

````python
expected = "=" * 30 + "\n이름: 김목포\n전공: 컴퓨터공학과\n목표: 매주 과제 제출하기\n" + "=" * 30
check("기본 카드 형식", lambda: make_card("김목포", "컴퓨터공학과", "매주 과제 제출하기") == expected)
check("앞뒤 공백 제거", lambda: make_card("  김목포 ", "컴퓨터공학과 ", " 매주 과제 제출하기") == expected)
check("정확히 5줄", lambda: len(make_card("가", "나", "다").split("\n")) == 5)
check("print가 아니라 return", lambda: isinstance(make_card("가", "나", "다"), str))
````

````python
# 결과 눈으로 확인하기
print(make_card("김목포", "컴퓨터공학과", "매주 과제 제출하기"))
````

## 준비: 자동 확인 도우미

````python
def check(desc, test):
    """자동 확인 도우미: test는 True/False를 돌려주는 함수(lambda)입니다."""
    try:
        ok = test()
    except NotImplementedError:
        print(f"⏳ 미작성 - {desc}")
    except Exception as e:
        print(f"❌ 오류 - {desc} ({type(e).__name__}: {e})")
    else:
        print(f"{'✅ 통과' if ok else '❌ 실패'} - {desc}")
````

## 문제 1. 학점 판정과 경계값 (2점)

`grade(score)`를 완성하세요.

| 점수 | 학점 |
|---|---|
| 95 이상 | `"A+"` |
| 90 이상 | `"A"` |
| 80 이상 | `"B"` |
| 70 이상 | `"C"` |
| 60 이상 | `"D"` |
| 60 미만 | `"F"` |
| 0 미만 또는 100 초과 | `"오류"` |

점수는 정수 또는 실수입니다. 그리고 **자동 확인 셀 아래에 여러분이 고른 경계값 테스트 4개를 직접 추가**하세요.

````python
def grade(score):
    """점수를 학점 문자열로 바꿉니다."""
    if score < 0 or score > 100:
        return "오류"
    if score >= 95:
        return "A+"
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 70:
        return "C"
    if score >= 60:
        return "D"
    return "F"
````

````python
table = {100: "A+", 95: "A+", 94.9: "A", 90: "A", 89.5: "B", 80: "B", 70: "C",
         69.99: "D", 60: "D", 59: "F", 0: "F", -0.1: "오류", 100.5: "오류"}
for s, want in table.items():
    check(f"grade({s}) == {want!r}", lambda s=s, want=want: grade(s) == want)
````

````python
# ✏️ 내가 고른 경계값 테스트 4개를 추가하세요. (예: check("grade(79.9)", lambda: grade(79.9) == "C"))
check("grade(79.9) == 'C'", lambda: grade(79.9) == "C")
check("grade(80.0) == 'B'", lambda: grade(80.0) == "B")
check("grade(60.0) == 'D'", lambda: grade(60.0) == "D")
check("grade(101) == '오류'", lambda: grade(101) == "오류")
````

## 문제 2. 목포대학 게임 (2점)

1부터 `n`까지 차례로 말하는 게임입니다. `mokpo_game(n)`은 말할 단어들을 **리스트**로 돌려줍니다.

- 3의 배수이면서 5의 배수: `"목포대학"`
- 3의 배수: `"목포"`
- 5의 배수: `"대학"`
- 나머지: 숫자를 문자열로 (`"1"`)

```python
mokpo_game(6)   # ["1", "2", "목포", "4", "대학", "목포"]
```

````python
def mokpo_game(n):
    """1부터 n까지의 목포대학 게임 결과를 리스트로 돌려줍니다."""
    words = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            words.append("목포대학")
        elif i % 3 == 0:
            words.append("목포")
        elif i % 5 == 0:
            words.append("대학")
        else:
            words.append(str(i))
    return words
````

````python
check("mokpo_game(6)", lambda: mokpo_game(6) == ["1", "2", "목포", "4", "대학", "목포"])
check("15번째는 '목포대학'", lambda: mokpo_game(15)[-1] == "목포대학")
check("길이가 n", lambda: len(mokpo_game(100)) == 100)
check("1~100 중 '목포대학' 6번", lambda: mokpo_game(100).count("목포대학") == 6)
check("1~100 중 '목포' 27번", lambda: mokpo_game(100).count("목포") == 27)
check("n=0이면 빈 리스트", lambda: mokpo_game(0) == [])
````

## 문제 3. 장학 판정과 집계 (3점)

### 3-1. `scholarship(gpa, credits, attendance)`

| 장학 | 조건 | 우선순위 |
|---|---|---|
| `"성적장학"` | 평점 4.0 이상 **그리고** 이수학점 15 이상 | 1 |
| `"성실장학"` | 출석률 95 이상 **그리고** 평점 3.0 이상 | 2 |
| `"근로장학"` | 이수학점 12 이상 **그리고** 출석률 80 이상 **그리고** 평점 2.0 이상 | 3 |
| `"해당없음"` | 그 외 | – |

여러 조건을 만족하면 **우선순위가 가장 높은 하나**만 돌려줍니다.

### 3-2. `count_scholarships(students)`

`(이름, 평점, 이수학점, 출석률)` 튜플의 리스트를 받아 **`(성적, 성실, 근로, 해당없음)` 인원 수 튜플**을 돌려줍니다.

````python
def scholarship(gpa, credits, attendance):
    """장학 종류를 돌려줍니다."""
    if gpa >= 4.0 and credits >= 15:
        return "성적장학"
    if attendance >= 95 and gpa >= 3.0:
        return "성실장학"
    if credits >= 12 and attendance >= 80 and gpa >= 2.0:
        return "근로장학"
    return "해당없음"


def count_scholarships(students):
    """(성적, 성실, 근로, 해당없음) 인원 수를 돌려줍니다."""
    merit = diligence = work = none = 0
    for _name, gpa, credits, attendance in students:
        kind = scholarship(gpa, credits, attendance)
        if kind == "성적장학":
            merit += 1
        elif kind == "성실장학":
            diligence += 1
        elif kind == "근로장학":
            work += 1
        else:
            none += 1
    return merit, diligence, work, none
````

````python
check("성적 경계 (4.0, 15)", lambda: scholarship(4.0, 15, 0) == "성적장학")
check("성적+성실 동시 만족 → 성적", lambda: scholarship(4.5, 20, 100) == "성적장학")
check("학점 부족 + 출석 우수 → 성실", lambda: scholarship(4.3, 12, 99) == "성실장학")
check("성실 경계 (3.0, 95)", lambda: scholarship(3.0, 0, 95) == "성실장학")
check("근로 경계 (2.0, 12, 80)", lambda: scholarship(2.0, 12, 80) == "근로장학")
check("근로 출석 부족 79.9", lambda: scholarship(3.5, 18, 79.9) == "해당없음")
check("평점 1.99", lambda: scholarship(1.99, 20, 100) == "해당없음")

sample = [("가", 4.2, 18, 90), ("나", 3.4, 17, 98), ("다", 2.5, 15, 85),
          ("라", 1.8, 19, 99), ("마", 4.1, 12, 70), ("바", 3.9, 21, 96)]
check("count_scholarships(sample) == (1, 2, 1, 2)", lambda: count_scholarships(sample) == (1, 2, 1, 2))
check("빈 목록 → (0, 0, 0, 0)", lambda: count_scholarships([]) == (0, 0, 0, 0))
````

## 문제 4. 저축 목표 계산기 – 무한 반복 막기 (AI 협업, 3점)

`months_to_goal(goal, monthly, rate=0.0, start=0)`은 목표 금액에 도달하는 데 걸리는 **개월 수**를 돌려줍니다.

매달 다음 순서로 계산합니다.

1. 기존 잔액에 월 이자를 붙인다: `balance = balance * (1 + rate)`
2. 월 저축액을 더한다: `balance += monthly`
3. 잔액이 `goal` **이상**이면 그달이 답이다.

**예외 규칙**

- 시작 잔액 `start`가 이미 `goal` 이상이면 `0`을 돌려줍니다.
- 600개월(50년) 안에 도달하지 못하면 `-1`을 돌려줍니다. (`monthly`가 0이고 이자도 없는 경우 등 → **무한 반복 방지**)

```python
months_to_goal(2_000_000, 300_000)            # 7
months_to_goal(1_000_000, 0, rate=0.01, start=500_000)  # 70
months_to_goal(1_000_000, 0)                  # -1
```

**AI 협업 과제**: AI에게 이 함수를 요청할 때 **무한 반복이 생길 수 있는 경우**를 프롬프트에 명시했는지, 받은 코드가 그 경우를 정말 처리하는지 확인하세요.

````python
def months_to_goal(goal, monthly, rate=0.0, start=0):
    """목표 금액 도달 개월 수. 이미 도달했으면 0, 600개월 안에 불가능하면 -1."""
    balance = start
    if balance >= goal:
        return 0
    month = 0
    while month < 600:
        month += 1
        balance = balance * (1 + rate)
        balance += monthly
        if balance >= goal:
            return month
    return -1
````

````python
check("이자 없이 30만원씩 → 7개월", lambda: months_to_goal(2_000_000, 300_000) == 7)
check("딱 나누어떨어짐 → 10개월", lambda: months_to_goal(1_000_000, 100_000) == 10)
check("이미 도달 → 0", lambda: months_to_goal(500_000, 100_000, start=500_000) == 0)
check("이자만으로 증가 → 70개월", lambda: months_to_goal(1_000_000, 0, rate=0.01, start=500_000) == 70)
check("저축 0, 이자 0 → -1 (무한 반복 방지)", lambda: months_to_goal(1_000_000, 0) == -1)
check("너무 적은 저축 → -1", lambda: months_to_goal(10**9, 1000) == -1)
check("이자 있으면 더 빨리", lambda: months_to_goal(10_000_000, 200_000, rate=0.005) < months_to_goal(10_000_000, 200_000))
````

## Level 2. 바꾸기 — 핵심 개념 연결

이제 `CSV 성적 파일 읽기와 검증`에 필요한 핵심 구조를 실행합니다.

1. 먼저 기본 코드를 실행합니다.
2. 입력값·조건·데이터 중 두 곳을 바꿉니다.
3. `assert` 또는 출력문으로 결과를 확인합니다.


```python
import csv
import io
import requests # Import the requests library

# The base URL for the GitHub repository datasets folder
github_base_url = "https://raw.githubusercontent.com/MNU-AI-Programming/AI-Assisted-Python-Programming/main/old/datasets/"
csv_filename = "scores.csv"

# Construct the full URL for the scores.csv file
file_url = f"{github_base_url}{csv_filename}"

try:
    # Fetch the content of the CSV file from the URL
    response = requests.get(file_url)
    response.raise_for_status() # Raise an exception for HTTP errors (4xx or 5xx)

    # Use io.StringIO to treat the fetched content as a file-like object
    file_content = io.StringIO(response.text)

    rows = list(csv.DictReader(file_content))
    scores = [int(row["score"]) for row in rows]
    print(f"학생 수: {len(rows)}, 평균: {sum(scores)/len(scores):.1f}")
    assert len(rows) >= 5
except requests.exceptions.RequestException as e:
    print(f"데이터를 불러오는 중 오류가 발생했습니다: {e}")
except KeyError as e:
    print(f"CSV 파일에 예상되는 컬럼이 없습니다: {e}")
except ValueError as e:
    print(f"점수 데이터를 숫자로 변환할 수 없습니다: {e}")
```

코드를 실행한 뒤 다음을 수행하세요.

1. 정상 입력의 예상 결과를 먼저 적습니다.
2. 경계값 또는 오류 가능 입력을 하나 추가합니다.
3. 실행 결과가 요구사항과 맞는지 테스트합니다.
4. 오류가 발생하면 메시지, 원인, 수정 내용을 기록합니다.


```python
def safe_int(text, default=0):
    try:
        return int(text)
    except (TypeError, ValueError):
        return default

values = [safe_int(x) for x in ["10", "오류", None, "25"]]
print(values)
assert values == [10, 0, 0, 25]
```
